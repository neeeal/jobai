# Scrape Jobs using JobSpy

## Defining Search Criteria

```python
## Static
SITE_NAME = ["indeed", "linkedin", "zip_recruiter", "google"] # "glassdoor", "bayt", "naukri", "bdjobs"
RESULTS_WANTED = 32
HOURS_OLD = 72

## Iterable (optional)
SEARCH_TERMS = ["software engineer", "chemical engineer"]
GOOGLE_SEARCH_TERMS = ["software engineer jobs near San Francisco, CA since yesterday"]
LOCATIONS = ["Philippines"]
COUNTRIES_INDEED = ['philippines']
```

## Job Search using JobSpy

### Single Search Sample

```python
## Single Search Sample

import csv
from jobspy import scrape_jobs

jobs = scrape_jobs(
    site_name=SITE_NAME,
    search_term=SEARCH_TERMS[0],
    google_search_term=GOOGLE_SEARCH_TERMS[0],
    location=LOCATIONS[0],
    results_wanted=RESULTS_WANTED,
    hours_old=HOURS_OLD,
    country_indeed=COUNTRIES_INDEED[0],
    
    linkedin_fetch_description=True, # gets more info such as description, direct job url (slower)
    # proxies=["208.195.175.46:65095", "208.195.175.45:65095", "localhost"],
)
print(f"Found {len(jobs)} jobs")
print(jobs.head())
jobs.to_csv("jobs.csv", quoting=csv.QUOTE_NONNUMERIC, escapechar="\\", index=False) # to_excel
```
